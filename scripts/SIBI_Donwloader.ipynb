```python
import asyncio
import nest_asyncio

nest_asyncio.apply()


async def cek_api():

    async with async_playwright() as p:

        browser = await p.chromium.launch(
            headless=True,
            args=[
                "--no-sandbox",
                "--disable-dev-shm-usage"
            ]
        )


        page = await browser.new_page()


        def lihat_request(req):

            url=req.url

            if (
                "api" in url.lower()
                or
                "book" in url.lower()
                or
                "catalog" in url.lower()
            ):

                print(
                    "REQUEST:",
                    url
                )


        page.on(
            "request",
            lihat_request
        )


        await page.goto(
            "https://buku.kemendikdasmen.go.id/katalog/buku-kurikulum-merdeka",
            wait_until="networkidle"
        )


        await page.wait_for_timeout(
            5000
        )


        await page.click(
            "#checkSD"
        )


        await page.wait_for_timeout(
            5000
        )


        await browser.close()



asyncio.get_event_loop().run_until_complete(
    cek_api()
)
```

```python
import requests
import os
import re

from tqdm.auto import tqdm
from concurrent.futures import ThreadPoolExecutor, as_completed


BASE_FOLDER = "Buku_SD_MI"


def clean_filename(name):

    name = re.sub(
        r'[\\/*?:"<>|]',
        "",
        name
    )

    return name.strip()



def download_file(row):

    jenis = row["jenis"]
    kelas = row["kelas"]


    folder = (
        f"{BASE_FOLDER}/"
        f"{jenis}/"
        f"Kelas {kelas}"
    )


    os.makedirs(
        folder,
        exist_ok=True
    )


    filename = clean_filename(
        row["judul"]
    ) + ".pdf"


    filepath = os.path.join(
        folder,
        filename
    )


    # skip jika sudah ada

    if os.path.exists(filepath):

        return "skip"



    try:

        r = requests.get(
            row["link_pdf"],
            timeout=120
        )


        if r.status_code == 200:

            with open(
                filepath,
                "wb"
            ) as f:

                f.write(
                    r.content
                )

            return "ok"


        return "failed"



    except Exception as e:

        return "error"



# ==========================
# MULTI DOWNLOAD
# ==========================


results=[]


with ThreadPoolExecutor(
    max_workers=8
) as executor:


    tasks = [
        executor.submit(
            download_file,
            row
        )
        for _,row in df.iterrows()
    ]


    for task in tqdm(
        as_completed(tasks),
        total=len(tasks)
    ):

        results.append(
            task.result()
        )



print(
    {
        x:results.count(x)
        for x in set(results)
    }
)
```
